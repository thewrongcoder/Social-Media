```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=students,
    x="social_media_hours",
    y="attention_span_minutes",
    hue="behavior_cluster"
)

plt.title("Student Behavioral Clusters")
plt.show()
```

```python
cluster_summary = students.groupby("behavior_cluster")[[
    "social_media_hours",
    "brain_rot_index",
    "attention_span_minutes",
    "study_hours_per_week"
]].mean()

cluster_summary
```

```python
cluster_means = cluster_summary.copy()

labels = {}

for cluster in cluster_means.index:

    sm = cluster_means.loc[cluster, "social_media_hours"]
    brain = cluster_means.loc[cluster, "brain_rot_index"]
    attention = cluster_means.loc[cluster, "attention_span_minutes"]

    if brain > cluster_means["brain_rot_index"].mean() and attention < cluster_means["attention_span_minutes"].mean():
        labels[cluster] = "High Risk"

    elif sm < cluster_means["social_media_hours"].mean() and attention > cluster_means["attention_span_minutes"].mean():
        labels[cluster] = "Focused"

    else:
        labels[cluster] = "Balanced"
```

```python
students["behavior_label"] = students["behavior_cluster"].map(labels)
```

Residual plots help visualize the errors (residuals) of a regression model. Ideally, residuals should be randomly scattered around zero, indicating that the model captures the underlying patterns well and there are no systematic biases. We will plot the residuals against the predicted values for both models.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate residuals for Linear Regression
residuals_lr = y_test - pred

# Calculate residuals for LightGBM
residuals_lgbm = y_test - y_pred_lgbm

plt.figure(figsize=(15, 6))

# Residual Plot for Linear Regression
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
sns.scatterplot(x=pred, y=residuals_lr, alpha=0.3, color='blue')
plt.axhline(y=0, color='red', linestyle='--')
plt.title('Linear Regression: Residuals vs Predicted Values')
plt.xlabel('Predicted Academic Risk Score (Linear Regression)')
plt.ylabel('Residuals (Actual - Predicted)')
plt.grid(True)

# Residual Plot for LightGBM
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
sns.scatterplot(x=y_pred_lgbm, y=residuals_lgbm, alpha=0.3, color='green')
plt.axhline(y=0, color='red', linestyle='--')
plt.title('LightGBM: Residuals vs Predicted Values')
plt.xlabel('Predicted Academic Risk Score (LightGBM)')
plt.ylabel('Residuals (Actual - Predicted)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

#### Quantitative Model Accuracy Comparison

To further compare the accuracy, we can look at common regression metrics:

*   **Mean Absolute Error (MAE):** The average of the absolute differences between predictions and actual values. It's less sensitive to outliers than MSE.
*   **Root Mean Squared Error (RMSE):** The square root of the Mean Squared Error (MSE). It is in the same units as the target variable, making it more interpretable than MSE.
*   **R-squared (R²):** The proportion of variance in the dependent variable that can be predicted from the independent variables. A higher R² indicates a better fit.

We will calculate these for both Linear Regression and LightGBM.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

print("--- Linear Regression Metrics ---")
mae_lr = mean_absolute_error(y_test, pred)
rmse_lr = np.sqrt(mean_squared_error(y_test, pred))
r2_lr = r2_score(y_test, pred) # Re-using previously calculated R2

print(f"MAE: {mae_lr:.4f}")
print(f"RMSE: {rmse_lr:.4f}")
print(f"R² Score: {r2_lr:.4f}")

print("\n--- LightGBM Metrics ---")
mae_lgbm = mean_absolute_error(y_test, y_pred_lgbm)
rmse_lgbm = np.sqrt(mean_squared_error(y_test, y_pred_lgbm)) # Re-using previously calculated MSE to get RMSE
r2_lgbm = r2_score(y_test, y_pred_lgbm) # Re-using previously calculated R2

print(f"MAE: {mae_lgbm:.4f}")
print(f"RMSE: {rmse_lgbm:.4f}")
print(f"R² Score: {r2_lgbm:.4f}")
```

### 23. TabPFN Implementation (all 500,000 rows, chunked)

TabPFN cannot run a single CPU fit on the full dataset (the training set is transformer context). This section still **uses every row** by splitting the data into chunks:

- **CPU:** 10,000-row chunks → 8,000 train / 2,000 test
- **GPU:** 20,000-row chunks → 16,000 train / 4,000 test

Each chunk uses the same 80/20 split (`random_state=42`). Test predictions are pooled so overall metrics cover a held-out 20% of all rows. Public TabPFN **v2** weights are used by default so this runs without a TabPFN-3 license token (`TABPFN_TOKEN`).


```python
import sys
!{sys.executable} -m pip install tabpfn
```

```python
import os

# Public v2 weights: TabPFN-3 checkpoints require TABPFN_TOKEN / license acceptance.
os.environ.setdefault("TABPFN_MODEL_VERSION", "v2")
os.environ["TABPFN_ALLOW_CPU_LARGE_DATASET"] = "true"

import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from scipy.stats import pearsonr
from tabpfn import TabPFNRegressor

features = [
    "social_media_hours",
    "brain_rot_index",
    "attention_span_minutes",
    "stress_level",
    "sleep_hours",
]
target = "academic_risk_score"

device = "cuda" if torch.cuda.is_available() else "cpu"
TEST_SIZE = 0.2
CHUNK_SIZE = 20_000 if device == "cuda" else 10_000
N_ESTIMATORS = 8 if device == "cuda" else 4

model_df = (
    students[features + [target]]
    .dropna()
    .sample(frac=1, random_state=42)
    .reset_index(drop=True)
)

print(f"PyTorch: {torch.__version__} | device={device}")
print(
    f"Using all {len(model_df):,} rows in chunks of {CHUNK_SIZE:,} "
    f"(n_estimators={N_ESTIMATORS})"
)


def report_r_metrics(split_name, y_true, y_pred):
    r, p_value = pearsonr(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    rmse = mean_squared_error(y_true, y_pred) ** 0.5
    mae = mean_absolute_error(y_true, y_pred)
    print(f"{split_name}")
    print(f"  R (Pearson): {r:.4f}  (p={p_value:.2e})")
    print(f"  R²:          {r2:.4f}")
    print(f"  RMSE:        {rmse:.4f}")
    print(f"  MAE:         {mae:.4f}")
    return r, r2


n = len(model_df)
n_chunks = (n + CHUNK_SIZE - 1) // CHUNK_SIZE
y_true_tabpfn = []
y_pred_tabpfn_chunks = []
y_pred_lr_tabpfn_chunks = []

for i in range(n_chunks):
    start = i * CHUNK_SIZE
    stop = min(start + CHUNK_SIZE, n)
    chunk = model_df.iloc[start:stop]
    if len(chunk) < 100:
        print(f"  skip tiny leftover chunk {i + 1}: {len(chunk)} rows")
        continue

    Xc = chunk[features]
    yc = chunk[target]
    X_tr, X_te, y_tr, y_te = train_test_split(
        Xc, yc, test_size=TEST_SIZE, random_state=42
    )

    tabpfn_model = TabPFNRegressor(
        n_estimators=N_ESTIMATORS,
        device=device,
        random_state=42,
        ignore_pretraining_limits=True,
        fit_mode="low_memory",
        memory_saving_mode=True,
    )
    tabpfn_model.fit(X_tr, y_tr)
    pred = np.asarray(tabpfn_model.predict(X_te))
    lr_pred = LinearRegression().fit(X_tr, y_tr).predict(X_te)

    y_true_tabpfn.append(np.asarray(y_te, dtype=float))
    y_pred_tabpfn_chunks.append(pred)
    y_pred_lr_tabpfn_chunks.append(np.asarray(lr_pred, dtype=float))

    r_chunk = pearsonr(y_te, pred)[0]
    print(
        f"  chunk {i + 1:02d}/{n_chunks}  rows={len(chunk):,}  "
        f"train={len(X_tr):,} test={len(X_te):,}  R={r_chunk:.4f}",
        flush=True,
    )

y_test_tabpfn = np.concatenate(y_true_tabpfn)
y_pred_tabpfn = np.concatenate(y_pred_tabpfn_chunks)
y_pred_lr_tabpfn = np.concatenate(y_pred_lr_tabpfn_chunks)

print(f"\nPooled TabPFN test predictions: {len(y_test_tabpfn):,} rows from {len(model_df):,} total")
print("\nTabPFN (chunked, all rows)")
r_test, r2_test = report_r_metrics("Test", y_test_tabpfn, y_pred_tabpfn)
print("\nLinearRegression (same TabPFN chunks)")
report_r_metrics("Test", y_test_tabpfn, y_pred_lr_tabpfn)
```

```python
plt.figure(figsize=(8, 6))
plot_n = min(len(y_test_tabpfn), 10_000)
rng = np.random.default_rng(42)
idx = rng.choice(len(y_test_tabpfn), size=plot_n, replace=False)
sns.regplot(
    x=y_pred_tabpfn[idx],
    y=y_test_tabpfn[idx],
    scatter_kws={"alpha": 0.35},
    line_kws={"color": "red"},
)
plt.xlabel("Predicted Academic Risk Score")
plt.ylabel("Actual Academic Risk Score (y_test)")
plt.title(
    f"TabPFN pooled test (all rows, chunked): Predicted vs Actual "
    f"(R={r_test:.3f}, R²={r2_test:.3f}; plot n={plot_n:,})"
)
plt.show()
```

### 23.5 LightGBM + XGBoost Hybrid

This hybrid averages LightGBM and XGBoost test-set predictions with equal weight (`0.5 / 0.5`). Both models already use the same features, target, and 80/20 split, so the blend is a simple ensemble of the two tree learners.

```python
import numpy as np
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split

if "y_pred_lgbm" not in globals() or "y_pred_xgb" not in globals():
    raise RuntimeError("Run the LightGBM and XGBoost cells first so y_pred_lgbm and y_pred_xgb exist.")

# Recreate the tree-model split so later TabPFN/TabICL cells cannot overwrite y_test.
_hybrid_features = [
    "social_media_hours",
    "brain_rot_index",
    "attention_span_minutes",
    "stress_level",
    "sleep_hours",
]
_, _, _, y_test_hybrid = train_test_split(
    students[_hybrid_features],
    students["academic_risk_score"],
    test_size=0.2,
    random_state=42,
)

y_pred_hybrid = 0.5 * np.asarray(y_pred_lgbm) + 0.5 * np.asarray(y_pred_xgb)

y_true = np.asarray(y_test_hybrid, dtype=float)
y_hat = np.asarray(y_pred_hybrid, dtype=float)
mse_hybrid = mean_squared_error(y_true, y_hat)
nonzero = y_true != 0
mape_hybrid = np.mean(np.abs((y_true[nonzero] - y_hat[nonzero]) / y_true[nonzero])) * 100

print("LightGBM + XGBoost Hybrid (equal-weight average)")
print(f"  R²:      {r2_score(y_true, y_hat):.4f}")
print(f"  MAE:     {mean_absolute_error(y_true, y_hat):.4f}")
print(f"  MSE:     {mse_hybrid:.4f}")
print(f"  RMSE:    {np.sqrt(mse_hybrid):.4f}")
print(f"  MAPE:    {mape_hybrid:.4f}%  (non-zero academic risk scores)")
```

## 24. TabICLv2 (all 500,000 rows, chunked)

TabICL is also an in-context learner, so a single CPU pass over 400,000 training rows is too heavy. This section uses **every row** with the same chunk protocol as TabPFN (10,000 on CPU / 20,000 on GPU, 80/20 inside each chunk). Test predictions are pooled for overall metrics.


```python
import sys

# Official TabICLv2 package (sklearn-style TabICLRegressor).
!{sys.executable} -m pip install tabicl
```

```python
import os

import numpy as np
import pandas as pd
import torch
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from tabicl import TabICLRegressor

features = [
    "social_media_hours",
    "brain_rot_index",
    "attention_span_minutes",
    "stress_level",
    "sleep_hours",
]
target = "academic_risk_score"

device = "cuda" if torch.cuda.is_available() else "cpu"
TEST_SIZE = 0.2
CHUNK_SIZE = 20_000 if device == "cuda" else 10_000
n_estimators = 8 if device == "cuda" else 1
batch_size = 4 if device == "cuda" else 1

model_df = (
    students[features + [target]]
    .dropna()
    .sample(frac=1, random_state=42)
    .reset_index(drop=True)
)

print(f"PyTorch: {torch.__version__} | device={device}")
print(f"TabICLv2 ensemble members: {n_estimators} | batch_size={batch_size}")
print(
    f"Using all {len(model_df):,} rows in chunks of {CHUNK_SIZE:,}"
)


def regression_metrics(y_true, y_hat):
    y_true = np.asarray(y_true, dtype=float)
    y_hat = np.asarray(y_hat, dtype=float)
    mse = mean_squared_error(y_true, y_hat)
    nonzero = y_true != 0
    mape = (
        np.mean(np.abs((y_true[nonzero] - y_hat[nonzero]) / y_true[nonzero])) * 100
        if nonzero.any()
        else np.nan
    )
    return {
        "R²": r2_score(y_true, y_hat),
        "MAE": mean_absolute_error(y_true, y_hat),
        "MSE": mse,
        "RMSE": float(np.sqrt(mse)),
        "MAPE (%)": mape,
    }


n = len(model_df)
n_chunks = (n + CHUNK_SIZE - 1) // CHUNK_SIZE
y_true_icl = []
y_pred_icl_chunks = []
y_pred_lr_icl_chunks = []

for i in range(n_chunks):
    start = i * CHUNK_SIZE
    stop = min(start + CHUNK_SIZE, n)
    chunk = model_df.iloc[start:stop]
    if len(chunk) < 100:
        print(f"  skip tiny leftover chunk {i + 1}: {len(chunk)} rows")
        continue

    Xc = chunk[features]
    yc = chunk[target]
    X_tr, X_te, y_tr, y_te = train_test_split(
        Xc, yc, test_size=TEST_SIZE, random_state=42
    )

    tabicl_model = TabICLRegressor(
        n_estimators=n_estimators,
        batch_size=batch_size,
        kv_cache=False,
        checkpoint_version="tabicl-regressor-v2-20260212.ckpt",
        device=device,
        offload_mode="auto",
        disk_offload_dir=os.path.join(os.getcwd(), "tabicl_offload"),
        random_state=42,
        verbose=False,
    )
    tabicl_model.fit(X_tr, y_tr)
    pred = np.asarray(tabicl_model.predict(X_te), dtype=float)
    lr_pred = LinearRegression().fit(X_tr, y_tr).predict(X_te)

    y_true_icl.append(np.asarray(y_te, dtype=float))
    y_pred_icl_chunks.append(pred)
    y_pred_lr_icl_chunks.append(np.asarray(lr_pred, dtype=float))

    r2_chunk = r2_score(y_te, pred)
    print(
        f"  chunk {i + 1:02d}/{n_chunks}  rows={len(chunk):,}  "
        f"train={len(X_tr):,} test={len(X_te):,}  R²={r2_chunk:.4f}",
        flush=True,
    )

y_test_icl = np.concatenate(y_true_icl)
y_pred_icl = np.concatenate(y_pred_icl_chunks)
y_pred_lr_icl = np.concatenate(y_pred_lr_icl_chunks)

tabicl_metrics = regression_metrics(y_test_icl, y_pred_icl)
metrics_table = pd.DataFrame(
    {
        "TabICLv2 (chunked)": tabicl_metrics,
        "Linear Regression (same chunks)": regression_metrics(y_test_icl, y_pred_lr_icl),
    }
).T
metrics_table.index.name = "Model"

print(f"\nPooled TabICLv2 test predictions: {len(y_test_icl):,} rows from {len(model_df):,} total")
print("\nTest-set metrics")
print(metrics_table.round(4).to_string())
try:
    display(metrics_table.style.format("{:.4f}"))
except NameError:
    pass
```
